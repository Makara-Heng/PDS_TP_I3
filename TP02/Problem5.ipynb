```python
# Step 1
prices = {"apple": 0.50, "banana": 0.25, "mango": 1.00}
print("Price of mango:", prices["mango"])
```

```python
# Step 2
prices["orange"] = 0.75
prices["apple"] = 0.60
del prices["banana"]
print("Prices:", prices)
```

```python
# Step 3
print("Price of grape:", prices.get("grape", "not sold"))
```

```python
# Step 4
for fruit, price in prices.items():
    print(f"{fruit:<7} ${price:.2f}")
```

```python
# step 5
basket = {"apple": 4, "mango": 2, "orange": 3}
total = 0
for fruit, qty in basket.items():
    total = total + prices[fruit] * qty
print(f"Total bill: ${total:.2f}")
```

# Answer
1. Dictionary better than two lists:
• With two lists, the fruit and its price are linked only by position (index). If you add, delete, or reorder one list and forget the other, prices silently go to the wrong fruit.
• A dictionary links each fruit directly to its price, so lookup is simple: prices[“mango”]. With lists you’d need prices_list[fruits_list.index(“mango”)], which searches item by item.

2. Difference between prices and prices.get:
• prices[“grape”] raises KeyError: ‘grape’ because the key doesn’t exist, and the program crashes.
• prices.get(“grape”) returns None if the key is missing, and prices.get(“grape”, “not sold”) returns the default “not sold”. The program keeps running.
• Use [] when the key must exist (a missing key means a bug). Use .get() when the key may be missing and you want to handle it gracefully.

3. Check the total by hand:
• 4 × 0.60 = 2.40
• 2 × 1.00 = 2.00
• 3 × 0.75 = 2.25
• Total = 2.40 + 2.00 + 2.25 = $6.65 ✔️ (matches the program output)
